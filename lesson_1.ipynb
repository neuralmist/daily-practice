```python
#load libraries 
import pandas as pd
```

```python
data={
    'region' : ['kerala','karnataka','kerala','tamilnadu','assam'],
    'patient_age' : [34,56,48,47,39],
    'hypertension_status':[0,1,1,0,1],
    'bmi':[21.7,29.5,33.4,24.5,25]
}
```

```python
df=pd.DataFrame(data)
```

```python
# filter a separate df for patients who have high hypertension status=1
high_hypertension = df[df["hypertension_status"]==1]
high_hypertension
```

```python
#mean of column oof high hypertension
bmi_mean = high_hypertension['bmi'].mean()
print(f"the average of bmi of high hypertension people are :{bmi_mean}")
```
