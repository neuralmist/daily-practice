```python
import pandas as pd
import numpy as np
```

```python
#messy data
messy_data={
    'patient_id':[201,202,203,204,205],
    'district':['Ernakulam','udupi','udupi','ernakulam','bangalore'],
    'glucose_level':[105,142,142,np.nan,98]
}

df=pd.DataFrame(messy_data)

print(df)
```

```python
#remove duplicate rows
df_cleaned = df.drop_duplicates()
df_cleaned
```

```python
df_cleaned["district"]=df_cleaned["district"].str.strip().str.title()
#it will capitalize all the first letter of values in the column
print(df_cleaned)
```

```python
median_glucose = df_cleaned["glucose_level"].median()
median_glucose
```

```python
df_cleaned["glucose_level"]=df_cleaned["glucose_level"].fillna(median_glucose)
print(df_cleaned)
```
