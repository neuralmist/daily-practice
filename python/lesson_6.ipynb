```python
import pandas as pd

#
case_data={
    'date':['2026-10-01','2026-10-02','2026-10-03','2026-10-04','2026-10-05'],
    'new_cases':[45,52,48,60,55]
}

df=pd.DataFrame(case_data)
df
```

```python
df.info()
```

```python
df["date"]=pd.to_datetime(df["date"])
df.info()
```

```python
#ranking data


df["severity_rank"] = df["new_cases"].rank(ascending=False)

print(df)
```

```python
# day change

df["yesterday_cases"]= df["new_cases"].shift(1)
df
```

```python
# changes in case

df["daily_change"]=df["new_cases"]-df["yesterday_cases"]

df
```
