```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
hr_data = {
    'patient_id':[10,11,12,13,14,15],
    'heart_rate':[72,75,350,68,80,74]

}

df=pd.DataFrame(hr_data)
df
```

```python
# box plot
sns.boxplot(x=df["heart_rate"],color="red")
plt.title("Heart rate (outlier)")
plt.xlabel("beats per min")
plt.show()
```

```python
# removing the outlier

df_clean=df[df["heart_rate"]<200]

df_clean
```
