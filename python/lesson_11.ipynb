```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
lifestyle_data={
    'patient_id':[601,602,603,604],
    'city':["Udupi","Manipal","Udupi","Mangalore"],
    'smoker':["Yes","No","No","Yes"]
}
```

```python
df=pd.DataFrame(lifestyle_data)
```

```python
# one hot encoding
#convert smoker column into binary indicator columns

df_encoded = pd.get_dummies(df,columns=["smoker"],dtype=int)
print(df_encoded)
```

```python
df_new =pd.get_dummies(df_encoded,columns=["city"],drop_first=True,dtype=int)
print(df_new)
# removes first city (drops mangalore. because of alphabetical order)
```

```python
df_new.columns
```

```python
# final df

final_df=pd.get_dummies(df,columns=["smoker","city"],dtype=int,drop_first=True)
final_df
```
