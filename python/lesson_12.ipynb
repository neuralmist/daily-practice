```python
#
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
claim_data={
    'patient_id':[701,702,703,704,705],
    'age':[25,45,35,60,22],
    'claim_amount':[1500,45000,12000,150000,800]
}

df=pd.DataFrame(claim_data)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
# normalization
# min max scalling
columns_to_be_scaled=["age","claim_amount"]

min_max_scaler=MinMaxScaler()

#fit & transform
df_min_max=df.copy()
df_min_max
```

```python
df_min_max[columns_to_be_scaled]=min_max_scaler.fit_transform(df[columns_to_be_scaled])
df_min_max
```

```python
# using standard scaling

std_scaler=StandardScaler()
df_standard=df.copy()

df_standard[columns_to_be_scaled]=std_scaler.fit_transform(df_standard[columns_to_be_scaled])
df_standard
```

```python
df_copy=df.copy()
std_scaler=StandardScaler()
df_copy["claim_amount"]=std_scaler.fit_transform(df_copy[["claim_amount"]])
df_copy
```
