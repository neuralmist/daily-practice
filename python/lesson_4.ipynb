# Dates and times

```python
import pandas as pd
import numpy as np
```

```python
admission_data={
    'patient_id':[301,302,303,304,305],
    'admission_date':['2025-11-15','2026-01-05','2026-03-22','2025-12-10','2026-06-18'],
    'diagnosis_code':['A1','B2','A1','C3','B2']
}
df=pd.DataFrame(admission_data)
print("--- data ----")
df
```

```python
#data types
df.info()
```

```python
#convert date to date type

df["admission_date"]=pd.to_datetime(df["admission_date"])
df.info()
```

```python
#to extract month and year from the date column

df["admit_year"]=df["admission_date"].dt.year

df["admit_month"]=df["admission_date"].dt.month

print("updated data frame with date month and year separate columns")
print(df)
```

```python
#filtering patient admitted based on date

admission_2026=df[df["admission_date"]>="2026-01-01"]
admission_2026
```

```python
#between march 01 - 31

admission_march = df[(df["admission_date"]>="2026-03-01") & (df["admission_date"]<="2026-03-31")]
admission_march
```
