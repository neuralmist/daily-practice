```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#table

biometric_data={
    'patient_id':[1,2,3,4,5,6,7,8],
    'age':[25,42,37,55,61,29,48,52],
    'bmi':[22.5,27.8,24.1,31.2,29.5,23.0,26.5,30.1],
    'diabetics_status':[0,1,0,1,1,0,0,1]

}

df=pd.DataFrame(biometric_data)
```

```python
#histogram

sns.histplot(data=df,x="age",bins=5,color="blue")
plt.title('distribution of age')
plt.xlabel("age--->")
plt.ylabel("count--->")

plt.show()
```

```python
#scatter plot
sns.scatterplot(data=df,x='age',y="bmi",color="red")
plt.xlabel("age ")
plt.ylabel("count-->")
plt.title("scatter plot")
plt.show()
```

```python
#scatterplot  using hue 

sns.scatterplot(data=df,x='age',y="bmi",hue="bmi")
plt.xlabel("age ---> ")
plt.ylabel("count-->")
plt.title("scatter plot")
plt.show()
```

```python

sns.scatterplot(data=df,x='age',y="bmi",hue="diabetics_status")
plt.xlabel("age ---> ")
plt.ylabel("count-->")
plt.title("scatter plot")
plt.show()
```
