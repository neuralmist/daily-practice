```python
import pandas as pd
import numpy as np
```

```python
trail_data={
    'patient_id':[501,502,503],
    'treatment':["Drug_A","Placebo","Drug_A"],
    'week_1_bp':[140,150,135],
    'week_2_bp':[135,148,130],
    'week_3_bp':[120,145,128]
}

df=pd.DataFrame(trail_data)
df
```

```python
#unpivot the dataset. using melt

df_long=pd.melt(
    df,
    id_vars=["patient_id","treatment"],
    value_vars=["week_1_bp","week_2_bp","week_3_bp"],
    var_name='measurement_week',
    value_name='blood_pressure'
)

print(df_long)
```

```python
#clean the test

df_long["measurement_week"]=df_long["measurement_week"].str.replace("_bp",'')
df_long
```

```python
#sort by patient_id/
df_long=df_long.sort_values(by=["patient_id","measurement_week"])
df_long
```
