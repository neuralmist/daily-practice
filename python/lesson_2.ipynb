```python
import pandas as pd
```

```python
demo_data= {
    'patient_id':[101,102,103,104,105],
    'state':['kerala','karnataka','maharashtra','kerala','tamil nadu'],
    'age':[45,52,34,28,61]
}

df_demo = pd.DataFrame(demo_data)
print(df_demo)
```

```python
clinical_data={
    'patient_id':[101,102,104,105],
    'hypertension_status':[1,1,0,1],
    "bmi":[26.5,29.0,22.1,31.4]
}
df_clinical=pd.DataFrame(clinical_data)
print(df_clinical)
```

```python
#inner merge
#merge dataset based on matching patient id

merge_inner = pd.merge(df_demo,df_clinical,on="patient_id",how="inner")

print("inner join table")
print(merge_inner)

```

```python
#left merge
merge_left=pd.merge(df_demo,df_clinical,on="patient_id",how="left")
print("left join")
print(merge_left)
```

```python
#right join

merge_right=pd.merge(df_demo,df_clinical,on="patient_id",how="right")
print("right merge table")
print(merge_right)
```

```python
# fill missing NaN with default 0
merge_left["hypertension_status"]=merge_left["hypertension_status"].fillna(0)
mean_bmi=merge_left["bmi"].mean()
merge_left["bmi"]=merge_left["bmi"].fillna(mean_bmi)
print("left merge table after filling Nan")
print(merge_left)
```

```python
#sort table who is more than 40 year old
merge_inner_ageabove40=merge_inner[merge_inner["age"]>40]
merge_inner_ageabove40
```

```python
#sort table who is in kerala
merge_table_kerala=merge_inner[merge_inner["state"]=="kerala"]
merge_table_kerala
```
