```python
import pandas as pd
```

```python
demographics=pd.DataFrame(
    {
        'patient_id':[901,902,903,904],
        'name':['Alice',"Rose","Oliver","Justin"]
    }
)

labs=pd.DataFrame({
    'patient_id':[901,902,903,905],
    "cholestrol":[180,220,195,240]
})

print(demographics)
```

```python
print(labs)
```

```python
#inner merge

inner_df=pd.merge(demographics,labs,on="patient_id",how="inner")
print(inner_df)
```

```python
#left merge

left_df=pd.merge(demographics,labs,on="patient_id",how="left")
print(left_df)
```

```python
#to fill the missing value
left_df["cholestrol"]=left_df["cholestrol"].fillna("not entered")
left_df
```

```python
#riht join
right_df=pd.merge(demographics,labs,on="patient_id",how="right")
right_df
```

```python
#outer join
outer_join=pd.merge(demographics,labs,on="patient_id",how="outer")
print("outer join","\n",outer_join)
```
