```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ehr_data={
    'patient_id':[801,802,803,804],
    'doctors_note':[
        'pt admitted for migrane.BP : 140/90. prescribed meds.',
        'Routine checkup . Bp is 110/70 today.No complaints.',
        'Patient has a history of severe asthma. bp 135/85.',
        'Follow-up visit. vitals normal.'
    ]

}

df=pd.DataFrame(ehr_data)
#note

for note in df["doctors_note"]:
    print(f"-{note}")
```

```python
#1. filter for patients who gave 'asthma. mentioned in their notes
asthma_patients= df[df["doctors_note"].str.contains("asthma",case=False)]
print(asthma_patients)
```

```python
#regex

#define pattern 
#bp patient level pattern
bp_pattern=r'(\d{2,3}/\d{2,3})'

#
df["extracted_bp"]=df["doctors_note"].str.extract(bp_pattern)


print("extracted_bp ----","\n",df[["patient_id","extracted_bp"]])
```

```python
#routine check
routine_patient= df[df["doctors_note"].str.contains("routine",case=True)]
print(routine_patient)
```

```python
routine_patient= df[df["doctors_note"].str.contains("routine",case=False)]
print(routine_patient)
```
