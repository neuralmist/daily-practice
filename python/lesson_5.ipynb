```python
import pandas as pd

treatment_data={
    'district':['Udupi','Udupi','Ernakulam','Ernakulam',"Udupi","Ernakulam"],
    'gender':['M','F','M','F','M','M'],
    'treatment_cost':[1200,1500,1300,1600,1150,1400],
    'recovery_days':[5,6,5,7,4,6]
}

df=pd.DataFrame(treatment_data)
print(df)
```

```python
#pivot matrix

cost_matrix=pd.pivot_table(
    df,values='treatment_cost',index='district',columns='gender',aggfunc="mean"
)

cost_matrix
```

```python
total_sum_and_avg_recovery=pd.pivot_table(
    df,
    index='district',
    aggfunc={'treatment_cost':'sum','recovery_days':'mean'},
    margins=True,
    margins_name='Grand Total'
)

print(total_sum_and_avg_recovery)
```
