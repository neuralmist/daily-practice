```python
import pandas as pd

clinic_data={
    'patient_id':[401,402,403,404,405],
    'age':[28,62,45,55,33],
    'fasting_sugar':[95,149,110,126,88]
}

df=pd.DataFrame(clinic_data)

df
```

```python
# .apply

def diagnose_sugar(sugar_level):
    if sugar_level<100:
        return 'Normal'
    elif 100<=sugar_level<=125:
        return 'Prediabetic'
    else:
        return 'Diabetic'


#  apply the function to the fasting_sugar column
df["diagnosis"]=df["fasting_sugar"].apply(diagnose_sugar)


print(df)
```

```python
# using lambda

df["risk_score"]=df["age"].apply(lambda x:x*1.5)


print(df)
```
