```python
#

import pandas as pd
```

```python
hospital_data={
    "patient_id":[1001,1002,1003,1004],
    "admit_date":["2026-10-01","2026-10-05","2026-10-15","2026-10-20"],
    "discharge_date":["2026-10-05","2026-10-12","2026-10-16","2026-10-30"]
}


df=pd.DataFrame(hospital_data)
print(df)
```

```python
df.dtypes
```

```python
# convert ttext column to date
df["admit_date"]=pd.to_datetime(df["admit_date"])
df["discharge_date"]=pd.to_datetime(df["discharge_date"])
print(df)
```

```python
print(df.dtypes)
```

```python
# calcualting the number of days between admit and discharge

df["number of days spent"]=(df["discharge_date"]-df["admit_date"])
print(df)
```

```python
df["number of days spent"]=(df["discharge_date"]-df["admit_date"]).dt.days
print(df)
```

```python
#follow up date
df["follow_up_date"]=(df["discharge_date"]+pd.Timedelta(days=14))
df
```
