```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
#FORMAT: dialect+driver://username:password@host:port/database_name
# if password as @   , use URL-encoded value for @ :-  %40
db_credentials="mysql+pymysql://root:Password%40123@localhost:3306/internship_prep"

engine=create_engine(db_credentials)
```

```python
query_1="""
select * from hospital_admissions
;
"""

df_sql_1=pd.read_sql(query_1,con=engine)
print(df_sql_1)
```

```python
query_2="""
select admission_date,diagnosis_code
from hospital_admissions
where admission_date > "2026-03-01";
"""

df_sql_2=pd.read_sql(query_2,con=engine)
print(df_sql_2)

print("data types:" ,"\n",df_sql_2.dtypes)
```

```python
# ssample data

predictions_data={
    "patient_id":[1005,1006,1007],
    "predicted_readmission":[1,0,1]

}

df_predicitions=pd.DataFrame(predictions_data)
df_predicitions
```
