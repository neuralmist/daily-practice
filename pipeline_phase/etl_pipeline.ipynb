```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
db_credentials="mysql+pymysql://root:Password%40123@localhost:3306/internship_prep"
engine=create_engine(db_credentials)
```

```python
messy_data={

    "patient_id":[1101,1102,1103,1104],
    "blood_type":["A+","b-","O+","O-"],  
    "cholesterol":[200,None,180,240]
}

pd.DataFrame(messy_data).to_sql("raw_lab_data",con=engine,if_exists="replace",index=False)
```

```python
query="select * from raw_lab_data"
df_raw=pd.read_sql(query,con=engine)
df_raw
```

```python
#2 . transform
# text format blood_type to upper
df_raw["blood_type"]=df_raw["blood_type"].str.upper()

df_raw
```

```python
avg_cholesterol = df_raw["cholesterol"].mean()
df_raw["cholesterol"] = df_raw["cholesterol"].fillna(avg_cholesterol)
df_raw
```

```python
# load

table_name="clean_lab_data"
df_raw.to_sql(table_name,con=engine,if_exists="replace",index=False)


query_1="show tables;"
df_1=pd.read_sql(query_1,con=engine)
```
