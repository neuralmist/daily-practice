```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("data/heart.csv")
```

```python
df.shape
```

```python
df.describe()
```

```python
df.columns
```

```python
# target variables
sns.set_theme(style="whitegrid")
plt.figure(figsize=(6,4))
sns.countplot(data=df,x="HeartDisease",palette="Set2")
plt.title("Distribution of heart disease")
plt.show()
```

```python
#gender
sns.set_theme(style="whitegrid")
plt.figure(figsize=(6,4))
sns.countplot(data=df,x="Sex",palette="Set3")
plt.title("Distribution of heart disease")
plt.show()
```

```python
# bivariate analysis

plt.figure(figsize=(10,6))

sns.boxplot(data=df,x="ChestPainType",y="Age",hue="HeartDisease",palette="coolwarm")
plt.title("age vs chest pain type by disease status")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x="Sex",y="Cholesterol",hue="HeartDisease")
plt.title("sex vs cholesterol  level and disease")
plt.show()
```
