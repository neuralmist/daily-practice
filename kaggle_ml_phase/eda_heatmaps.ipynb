```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
data={
    "age":[25,45,35,60,22,55,68,42,38,50],
    "bmi":[22.1,28.5,25.0,31.2,21.0,29.5,32.1,26.8,24.5,27.9],
    "blood_pressure":[110,135,120,150,105,140,155,125,118,130],
    "cholestrol":[180,220,195,250,175,240,260,210,190,225]
}

df=pd.DataFrame(data)
df
```

```python
# correlation matrix
correlation_matrix=df.corr()

print("correlation matrix:","\n",correlation_matrix)
```

```python
# plot

plt.figure(figsize=(8,6))

#heatmap
# annot = True , will print the number inside the boxes
# cmap="coolwarm"  uses blue for negative correlation and red for positive
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)
plt.title(" clincial feature correlation heatmap")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix,annot=True,cmap="viridis",linewidths=1)
plt.title("Feature correlation heatmap (viriis)")
plt.tight_layout()
plt.show()
```
