```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("data/RELIANCE.csv",parse_dates=["Date"],index_col="Date").sort_index()
df.head()
```

```python
df.isna().sum().sum()
```

```python
df.columns
```

```python
# new column Daily return
df["Daily_Returns_%"]=df["Close"].pct_change()*100
print("-- raw data vs stationary data --")
print(df[["Close","Daily_Returns_%"]].head())
```

```python
print(df[["Close","Daily_Returns_%"]].tail())
```

```python
# calculating 20 day historical volatility
df["Volatility_20d"]=df["Daily_Returns_%"].rolling(window=20).std()

print("--Recent volatility=--")
print(df[["Daily_Returns_%","Volatility_20d"]].tail(10))
```

```python
# target next day return
df["Target_Next_Day_Returns_%"]=df["Daily_Returns_%"].shift(-1)

# target direction (binary target)
df["Target_Direction"]=np.where(df["Target_Next_Day_Returns_%"]>0,1,0)


print(df[["Daily_Returns_%","Target_Next_Day_Returns_%","Target_Direction"]].tail(10))
```
