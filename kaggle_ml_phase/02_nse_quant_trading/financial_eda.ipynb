```python
print("total trading days:",len(df))
```

```python
df[["Open","High","Low","Close","Volume"]].head()
```

```python
# calculating Simple MOving Average (SMA) using rolling windows
# windows =50   means look back at the previous 50 trading days and calcualte the mean
df['SMA_50'] = df['Close'].rolling(window=50).mean()
df['SMA_200'] = df['Close'].rolling(window=200).mean()

print(df[['Close', 'SMA_50', 'SMA_200']].tail())
```

```python
sns.set_theme(style="darkgrid")
plt.figure(figsize=(14, 7))

recent_data = df.loc['2020-01-01':]

sns.lineplot(data=recent_data,x=recent_data.index, y='Close',label='Reliance close price', color='blue', alpha=0.4)
sns.lineplot(data=recent_data,x=recent_data.index, y='SMA_50',label='50-DaySMA', color='orange', linewidth=2)
sns.lineplot(data=recent_data,x=recent_data.index, y='SMA_200',label='200-Day SMA', color='red', linewidth=2)

plt.title("RelianceIndustries (RELIANCE)-price trend and moving averages", fontsize=16)
plt.xlabel("Date")
plt.ylabel("Price (INR)")
plt.legend()
plt.tight_layout()
plt.show()
```
