```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
```

```python
data={
    "age":[45,22,50,60,35,70,48,55],
    "blood_pressure":[120,110,140,150,115,160,135,145],
    "smoker":["Yes","No","Yes","No","No","Yes","No","Yes"],
    "diabetes":[1,0,1,1,0,1,0,1]
}

df=pd.DataFrame(data)


df
```

```python
# split x ,y

x=df[["age","blood_pressure","smoker"]]
y=df["diabetes"]
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.3)
x_train.shape
```

```python
x_test.shape
```

```python
# columntransformer

numeric_cols=["age","blood_pressure"]
categorical_cols=["smoker"]


 # column transformer
preprocessor=ColumnTransformer(
    transformers=[
        ("num_scaler",StandardScaler(),numeric_cols),
        ("cat_encoder",OneHotEncoder(drop="first"),categorical_cols)
    ]
)
```

```python
# ML pipelin

ml_pipeline=Pipeline(
    steps=[
        ("preprossing",preprocessor),
        ("classifier",RandomForestClassifier(random_state=42))
    ]
)

ml_pipeline.fit(x_train,y_train)

prediction=ml_pipeline.predict(x_test)

print(f"predcitions: {prediction}")
print(f"actual value:{y_test.values}")
```

```python
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(prediction,y_test)
print("accuracy :",accuracy*100)
```
