```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python

data = {
    'age':[25,42,37,55,61,29,48,52,33,40],
    'bmi':[22.5,27.8,24.1,31.2,29.5,23.0,26.5,30.1,25.0,28.5],
    'blood_sugar':[90,140,110,150,160,95,105,145,100,130],
    "diabetes_status":[0,1,0,1,1,0,0,1,0,1]  # 1-diabetic,0-normal
}

df= pd.DataFrame(data)
print(df)
```

```python
#Train test split

x=df[["age","bmi","blood_sugar"]]
y=df["diabetes_status"]
x.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
print(x_train)

```

```python
model = LogisticRegression()

model.fit(x_train,y_train)

predictions=model.predict(x_test)
print("predicted_outcomes",predictions)
print("y_test",y_test.values)
```

```python

accuracy=accuracy_score(y_test,predictions)
print("accuracy_score :",accuracy)
```
